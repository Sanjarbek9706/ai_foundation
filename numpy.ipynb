```python
import numpy as np

arr = np.array([1, 3, 5, 7])
print("numpy_array:", arr)

arr.shape
```

```python

# shape
arr2 = np.array([[1, 3, 5], [2, 5, 1]])
print("array 2", arr2)
arr2.shape
```

```python

# dtype()
a = np.array([1, 2, 3])
print(a.dtype)
```

```python

# reshape()

a = np.array([1, 2, 3, 4, 5, 6])
b = a.reshape(2, 3)
print("shaped_array", b)
```

```python

# astype()

a = np.array([1, 2, 3, 4, 5, 6])
b = a.astype(float)
print("float", b)
```

```python

# np.zeros(), np.ones() -> dummy data

z = np.zeros([2, 3])
o = np.ones([2, 3])
print("zeros:", z)
print("ones:", o)
```

```python
# ndim

n = np.array([[1, 2, 3, 4], [4,3,2,1]])
print("ndim:", n.ndim)
```

```python
# arange()
# Malumotlar kamlik qilganda, bu metod orqalik test data ni yaratib olishimiz mumkun.

a = np.arange(1, 6)
print("arrange->", a)
```

```python
# sum(), mean(), max(), min()

a = np.array([1, 2, 3, 4, 5 ])
print("sum->", a.sum())
print("mean->", a.mean())
print("max->", a.max())
print("min->", a.min())
```

```python
# flatten(), bu method 2d, 3d arrayni 1d list ga aylatirib beradi

a = np.array([[1, 2, 3], [3, 2, 1]])
print("flatten->", a.flatten())
```

```python
# T - (transpose)
# array shaklini o'zgartirish talab qilinganda, yoki malumotlarni qayta tashkilashda.

a = np.array([[1, 2, 3], [4, 5, 6]])
print("transpose: \n", a.T)
```

```python
# dot() & @ - matrisalarni kopaytirish payitida ishlatamiz

a = np.array([[1, 2], [3, 4]])
b = np.array([[2, 0], [1, 6]])
print("dot/@ ->", a.dot(b))
```

```python
# random.rand() - array yaritish vaqtida random raqamlarni yaratib beradi
# 0.0 -> 1.0 (0 dan 1 gacha bolgan sonlar oralig'ida)
# [0.0, 1)
a = np.random.rand(2, 3)
print("random rand->", a)
```
